### Bài 2. Tạo các Tensor thường dùng

```python
import torch
a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)
print(a)
print(b)
print(c)
print(d)
print(e)
```

#### Sự khác nhau giữa torch.rand() và torch.randn():
torch.rand(*size):
- Phân phối: Phân phối đều.
- Khoảng giá trị: Các số ngẫu nhiên được sinh ra nằm trong khoảng đều từ [0, 1) (từ 0 đến gần bằng 1).
- Ứng dụng: Thường dùng khi bạn cần khởi tạo các giá trị ngẫu nhiên có trọng số hoặc tỉ lệ phân bố đồng đều trong một khoảng giới hạn.
torch.randn(*size):
- Phân phối: Phân phối chuẩn.
- Khoảng giá trị: Các số ngẫu nhiên có giá trị trung bình (mean) bằng 0 và độ lệch chuẩn (standard deviation) bằng 1. Các giá trị tập trung nhiều quanh số 0, có thể có giá trị âm hoặc dương (vượt quá khoảng $[0, 1)$).
- Ứng dụng: Rất phổ biến trong Deep Learning để khởi tạo trọng số (weights) cho mạng neural hoặc tạo nhiễu.

### Bài 3. Kiểm tra thông tin Tensor

```python
import torch
a = torch.zeros(2, 3, 4)

print("Tensor:")
print(a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

Trong PyTorch, torch.Size([2, 3, 4]) đại diện cho kích thước (shape) của một Tensor 3 chiều (3D Tensor).
- Chiều 0 (Dimension 0 - Trục 0): Có 2 mảng/ma trận 2D ở lớp ngoài cùng.
- Chiều 1 (Dimension 1 - Trục 1): Mỗi ma trận 2D đó bao gồm 3 dòng (rows).
- Chiều 2 (Dimension 2 - Trục 2): Mỗi dòng chứa 4 cột (columns).

### Bài 4. Slicing và Indexing

```python
import torch
a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
```

Kích thước (Shape) của các Tensor con: 
- Shape của a[1]: torch.Size([4, 5])
- Shape của a[1:, 2:4]: torch.Size([2, 2, 5])

### Thao tác lấy hai phần tử cuối cùng trên chiều thứ ba

```python
sub_a = a[:, :, -2:]

print("Tensor mới chứa 2 phần tử cuối trên chiều thứ ba:")
print(sub_a)
print("Shape mới:", sub_a.shape)
```

### Bài 5. Reshape và Flatten

```python
import torch
a = torch.randn(3, 4)

print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
```

### Kiểm tra số phần tử trước và sau khi reshape

```python
num_before = a.numel()
num_after_ravel = a.ravel().numel()
num_after_reshape = a.reshape(3, 2, 2).numel()

print(f"Số phần tử ban đầu (3x4): {num_before}")
print(f"Số phần tử sau khi ravel(): {num_after_ravel}")
print(f"Số phần tử sau khi reshape(3, 2, 2): {num_after_reshape}")
```

### Thử reshape Tensor thành kích thước 2 × 6

```python
b = a.reshape(2, 6)

print("Tensor sau khi reshape thành 2 x 6:")
print(b)
print("Shape mới:", b.shape)
```

### Bài 6. Phép toán và nhân ma trận

```python
import torch
a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)
print("b:")
print(b)
print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\nMatrix multiplication:")
print(a @ b.T)
```

Phân biệt phép toán a * b và a @ b.T:
Phép toán a * b:
- Cơ chế: Nhân từng phần tử tương ứng (element-by-element) của Tensor a với Tensor b.
- Yêu cầu: a và b phải cùng kích thước.
- Kết quả: Trả về một Tensor mới có cùng kích thước ban đầu.
Phép toán a @ b.T:
- Cơ chế: Phép nhân ma trận đại số tuyến tính.
- Yêu cầu: Số cột của ma trận thứ nhất phải bằng số hàng của ma trận thứ hai. Với a (2 x 3) và b.T (3 x 2), phép nhân hợp lệ vì chiều trong trùng nhau (3 = 3).
- Kết quả: Trả về một ma trận tích theo quy tắc đại số.

Kích thước (Shape) của kết quả a @ b.T:
- Shape kết quả: torch.Size([2, 2])

### Bài 7. Các hàm thống kê trên Tensor

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))
```

Thay dim=0 bằng dim=1 và nhận xét sự khác nhau:
Khi dùng dim=0:
- Hướng thực thi: Thao tác dọc (chiều dọc theo 3 hàng).
- Giá trị trung bình (mean): Tính trung bình của từng cột.
- Độ lệch chuẩn (std): Tính độ lệch chuẩn của từng cột.
- Tổng tích lũy (cumsum): Cộng dồn từ hàng 0 xuống hàng 2.
Khi dùng dim=1:
- Hướng thực thi: Thao tác ngang (chiều ngang theo 4 cột).
- Giá trị trung bình (mean): Tính trung bình của từng hàng.
- Độ lệch chuẩn (std): Tính độ lệch chuẩn của từng hàng.
- Tổng tích lũy (cumsum): Cộng dồn từ cột 0 sang cột 3.
