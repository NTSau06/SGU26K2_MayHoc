#### Sự khác nhau giữa torch.rand() và torch.randn():
torch.rand(*size):
- Phân phối: Phân phối đều.
- Khoảng giá trị: Các số ngẫu nhiên được sinh ra nằm trong khoảng đều từ [0, 1) (từ 0 đến gần bằng 1).
- Ứng dụng: Thường dùng khi bạn cần khởi tạo các giá trị ngẫu nhiên có trọng số hoặc tỉ lệ phân bố đồng đều trong một khoảng giới hạn.
torch.randn(*size):
- Phân phối: Phân phối chuẩn.
- Khoảng giá trị: Các số ngẫu nhiên có giá trị trung bình (mean) bằng 0 và độ lệch chuẩn (standard deviation) bằng 1. Các giá trị tập trung nhiều quanh số 0, có thể có giá trị âm hoặc dương (vượt quá khoảng $[0, 1)$).
- Ứng dụng: Rất phổ biến trong Deep Learning để khởi tạo trọng số (weights) cho mạng neural hoặc tạo nhiễu.

### Bài 6. Phép toán và nhân ma trận

Phân biệt phép toán a * b và a @ b.T:
Phép toán a * b:

- Yêu cầu: Số cột của ma trận thứ nhất phải bằng số hàng của ma trận thứ hai. Với a (2 x 3) và b.T (3 x 2), phép nhân hợp lệ vì chiều trong trùng nhau (3 = 3).
- Kết quả: Trả về một ma trận tích theo quy tắc đại số.

- Shape kết quả: torch.Size([2, 2])

### Bài 7. Các hàm thống kê trên Tensor

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))
```

Thay dim=0 bằng dim=1 và nhận xét sự khác nhau:
Khi dùng dim=0:
- Hướng thực thi: Thao tác dọc (chiều dọc theo 3 hàng).
- Giá trị trung bình (mean): Tính trung bình của từng cột.
- Độ lệch chuẩn (std): Tính độ lệch chuẩn của từng cột.
- Tổng tích lũy (cumsum): Cộng dồn từ hàng 0 xuống hàng 2.
Khi dùng dim=1:
- Hướng thực thi: Thao tác ngang (chiều ngang theo 4 cột).
- Giá trị trung bình (mean): Tính trung bình của từng hàng.
- Độ lệch chuẩn (std): Tính độ lệch chuẩn của từng hàng.
- Tổng tích lũy (cumsum): Cộng dồn từ cột 0 sang cột 3.

### Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

So sánh ba hệ số học được với [1, 2, 3]:
- Ba hệ số học được trong w sẽ gần như trùng khớp hoàn toàn với [[1.], [2.], [3.]] (hoặc chênh lệch cực kỳ nhỏ do sai số số thực float32).
Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step():
- optimizer.zero_grad(): Xóa các giá trị đạo hàm (gradient) đã lưu ở vòng lặp trước trong biến w.grad. Trong PyTorch, gradient mặc định sẽ cộng dồn (accumulate) qua mỗi lần gọi .backward(). Nếu không xóa, gradient mới sẽ cộng thêm vào gradient cũ khiến bước cập nhật trọng số bị sai hướng.
- mse.backward(): Thực hiện thuật toán lan truyền ngược (Backpropagation). Lệnh này tính đạo hàm riêng của hàm mất mát mse theo biến trọng số w và lưu kết quả đó vào thuộc tính w.grad.
- optimizer.step(): Thực hiện cập nhật trọng số w dựa trên gradient vừa tính được (w.grad) và thuật toán tối ưu NAdam cùng với tốc độ học lr=0.01.

### Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)

print("C =", C)
print("D =", D)
```

### Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.

```python
import random
import torch

# 1. Khởi tạo ngẫu nhiên
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

# 2. Huấn luyện bằng NAdam
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1 * y1 + y2 * y2 + y3 * y3 + y4 * y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

# Lấy giá trị float từ tensor
a, b, c, d = A.item(), B.item(), C.item(), D.item()

# 3. In giá trị tìm được
print("=== NGHIỆM TÌM ĐƯỢC ===")
print(f"A = {a:.4f}, B = {b:.4f}, C = {c:.4f}, D = {d:.4f}\n")

# 4. Kiểm tra mức độ thỏa mãn các phương trình
print("=== KIỂM TRA VỚI 4 PHƯƠNG TRÌNH ===")
print(f"1) A + B = {a + b:.4f} (Mục tiêu: 9)")
print(f"2) C - D = {c - d:.4f} (Mục tiêu: 1)")
print(f"3) A + C = {a + c:.4f} (Mục tiêu: 8)")
print(f"4) B - D = {b - d:.4f} (Mục tiêu: 2)")

# 5. Kiểm tra tính đúng đắn theo dạng nghiệm tổng quát (7-t, t+2, t+1, t)
t = d
print("\n=== ĐỐI CHIẾU VỚI CÔNG THỨC NGHIỆM TỔNG QUÁT ===")
print(f"Nghiệm thực tế    : ({a:.4f}, {b:.4f}, {c:.4f}, {d:.4f})")
print(f"Nghiệm theo lý thuyết: ({7-t:.4f}, {t+2:.4f}, {t+1:.4f}, {t:.4f})")
```

#### Phép toán trên Tensor A và B (3 × 4)

```python
import torch

torch.manual_seed(42)  # Cố định ngẫu nhiên để dễ đối chiếu
A = torch.randn(3, 4)
B = torch.randn(3, 4)

# Các phép toán phần tử (element-wise)
A_plus_B = A + B
A_mul_B = A * B
A_div_B = A / B

# Trung bình và độ lệch chuẩn theo cột (dim=0)
mean_A_col = torch.mean(A, dim=0)
std_A_col = torch.std(A, dim=0)

mean_B_col = torch.mean(B, dim=0)
std_B_col = torch.std(B, dim=0)

print("A + B:\n", A_plus_B)
print("\nA * B:\n", A_mul_B)
print("\nA / B:\n", A_div_B)
print("\nA - Trung bình theo cột:", mean_A_col)
print("A - Độ lệch chuẩn theo cột:", std_A_col)
print("\nB - Trung bình theo cột:", mean_B_col)
print("B - Độ lệch chuẩn theo cột:", std_B_col)
```

#### Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import numpy as np
import torch

# 1. Tạo dữ liệu giả lập từ y = 2x^2 - 3x + 5
polynomial = np.poly1d([2, -3, 5])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

# 2. Tạo ma trận đặc trưng [X^2, X, 1]
XX = np.hstack([X * X, X, np.ones_like(X)])

# 3. Khởi tạo trọng số ngẫu nhiên
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

# 4. Huấn luyện với NAdam
optimizer = torch.optim.NAdam([w], lr=0.01)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Các hệ số tìm được (Learned coefficients):")
print(w.detach().numpy())
```
