### Bài 1. Tạo Tensor cơ bản

```python
import torch
a = torch.tensor(
[[1, 2, 3],
[4, 5, 6]],
dtype=torch.int32
)
print(a)
```

### Bài 2. Tạo các Tensor thường dùng

```python
import torch
a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)
print(a)
print(b)
print(c)
print(d)
print(e)
```

#### Sự khác nhau giữa torch.rand() và torch.randn():
torch.rand(*size):
- Phân phối: Phân phối đều.
- Khoảng giá trị: Các số ngẫu nhiên được sinh ra nằm trong khoảng đều từ [0, 1) (từ 0 đến gần bằng 1).
- Ứng dụng: Thường dùng khi bạn cần khởi tạo các giá trị ngẫu nhiên có trọng số hoặc tỉ lệ phân bố đồng đều trong một khoảng giới hạn.
torch.randn(*size):
- Phân phối: Phân phối chuẩn.
- Khoảng giá trị: Các số ngẫu nhiên có giá trị trung bình (mean) bằng 0 và độ lệch chuẩn (standard deviation) bằng 1. Các giá trị tập trung nhiều quanh số 0, có thể có giá trị âm hoặc dương (vượt quá khoảng $[0, 1)$).
- Ứng dụng: Rất phổ biến trong Deep Learning để khởi tạo trọng số (weights) cho mạng neural hoặc tạo nhiễu.

### Bài 3. Kiểm tra thông tin Tensor

```python
import torch
a = torch.zeros(2, 3, 4)

print("Tensor:")
print(a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

Trong PyTorch, torch.Size([2, 3, 4]) đại diện cho kích thước (shape) của một Tensor 3 chiều (3D Tensor).
- Chiều 0 (Dimension 0 - Trục 0): Có 2 mảng/ma trận 2D ở lớp ngoài cùng.
- Chiều 1 (Dimension 1 - Trục 1): Mỗi ma trận 2D đó bao gồm 3 dòng (rows).
- Chiều 2 (Dimension 2 - Trục 2): Mỗi dòng chứa 4 cột (columns).

### Bài 4. Slicing và Indexing

```python
import torch
a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
```

Kích thước (Shape) của các Tensor con: 
- Shape của a[1]: torch.Size([4, 5])
- Shape của a[1:, 2:4]: torch.Size([2, 2, 5])

### Thao tác lấy hai phần tử cuối cùng trên chiều thứ ba

```python
sub_a = a[:, :, -2:]

print("Tensor mới chứa 2 phần tử cuối trên chiều thứ ba:")
print(sub_a)
print("Shape mới:", sub_a.shape)
```

### Bài 5. Reshape và Flatten

```python
import torch
a = torch.randn(3, 4)

print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
```

### Kiểm tra số phần tử trước và sau khi reshape

```python
num_before = a.numel()
num_after_ravel = a.ravel().numel()
num_after_reshape = a.reshape(3, 2, 2).numel()

print(f"Số phần tử ban đầu (3x4): {num_before}")
print(f"Số phần tử sau khi ravel(): {num_after_ravel}")
print(f"Số phần tử sau khi reshape(3, 2, 2): {num_after_reshape}")
```

### Thử reshape Tensor thành kích thước 2 × 6

```python
b = a.reshape(2, 6)

print("Tensor sau khi reshape thành 2 x 6:")
print(b)
print("Shape mới:", b.shape)
```
